```python
"""
Oppgave 5

Arealet og omkrets av en figur bestående av en trekant og en halvsirkel, der a er både den 
hosliggende katet i den rettvinklede trekanten og radius i halvsirkelen, og b er motstående katet i trekanten
"""

import numpy as np
import math

def fun_figur(a, b):

    areal_figur = ((a * b)/2)+((np.pi * (a/2)**2)/2) #Arealet av trekanten og halvssirkel
    omkrets_figur = ((np.pi * (a/2)) + (math.sqrt(a**2 + b**2) + b)) #Omkretsen av buen i halvsirkelen og trekanten utenom siden a

    return areal_figur, omkrets_figur


a = float(input("Skriv inn a:"))
b = float(input("Skriv inn b:"))


areal_figur, omkrets_figur = fun_figur(a, b)


print("Arealet av figuren er:",areal_figur)
print("Omkretsen av figuren er:",omkrets_figur)

```

```python
"""
Oppgave 6

Kode som plotter funksjonen 𝑓(𝑥) = −𝑥2 − 5, for x på intervallet [-10,10]
"""
import numpy as np
import matplotlib.pyplot as plt

def fun_plot(x):
    y = x**2 - 5
    return (y)

x= np.linspace(-10, 10, 200) 
y = fun_plot(x)
    
    
plt.plot(x, y)
plt.grid()
plt.title("Plot for X over intervallet -10 til 10")
plt.xlabel("x")
plt.ylabel("Verdien til funksjonen f når x er gitt")
plt.show()
```
